```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Handling missing values
# Since there are no missing values, this step confirms the dataset is already clean in this part.

df = df.dropna()

print("Missing values after handling:")
print(df.isnull().sum())

print("\nDataset shape after handling missing values:", df.shape)
```

```python
print("Rows before removing duplicates:", df.shape[0])

df = df.drop_duplicates()

print("Rows after removing duplicates:", df.shape[0])
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
# Filtering records where Sales is greater than 100

filtered_df = df[df["Sales"] > 100]

print("Original dataset shape:", df.shape)
print("Filtered dataset shape where Sales > 100:", filtered_df.shape)

filtered_df.head()
```

```python
# Selecting important columns from the dataset

selected_columns = df[[
    "Order ID",
    "Order Date",
    "Customer Name",
    "Category",
    "Sub-Category",
    "Product Name",
    "Sales",
    "Quantity",
    "Profit"
]]

print("Selected columns shape:", selected_columns.shape)

selected_columns.head()
```

```python
# Creating a new column: total_amount = price * quantity

df["Price"] = df["Sales"] / df["Quantity"]

df["total_amount"] = df["Price"] * df["Quantity"]

df[["Product Name", "Sales", "Quantity", "Price", "total_amount"]].head()
```

```python
# Saving the cleaned dataset as a new CSV file

df.to_csv("cleaned_superstore.csv", index=False)

print("Cleaned dataset saved successfully as cleaned_superstore.csv")
print("Final dataset shape:", df.shape)
```
